#Task 4 — Data Wrangling

Step 1: Splitting the Dataframe

```python
# User-related info
df_users = df[['user_id','age','gender','country','follower_count','is_verified']]

# Post-related info
df_posts = df[['user_id','post_id','post_type','post_category','likes','comments','shares','impression_count','sentiment','hashtags','engagement_rate']]
```

Step 2: Merging Function

```python
# Merge on user_id
merged_df = pd.merge(df_users, df_posts, on="user_id", how="inner")
merged_df
```

Join Function

```python
# Set index to user_id
df_users = df_users.set_index('user_id')
df_posts = df_posts.set_index('user_id')

# Join by index
joined_df = df_users.join(df_posts, how="inner")
joined_df
```

Concat Function

```python
# Vertical concat (stack rows)
df1 = df.iloc[:3]   # first 3 rows
df2 = df.iloc[3:]   # remaining rows
concat_vertical = pd.concat([df1, df2], axis=0)
concat_vertical
```

```python
# Horizontal concat (stack columns)
concat_horizontal = pd.concat([df_users, df_posts], axis=1)
concat_horizontal
```

Wrangling Tasks

```python
# Engagement score
df['engagement_score'] = df['likes'] + df['comments'] + df['shares']

# Log-transformed metrics
df['log_likes'] = np.log1p(df['likes'])
df['log_impressions'] = np.log1p(df['impression_count'])

# Hashtag count
df['hashtag_count'] = df['hashtags'].fillna("").apply(lambda x: len(str(x).split()))

df.head(2)
```

```python
# Total impressions by country
df.groupby('country')['impression_count'].sum()
```

```python
# Average engagement rate by sentiment
df.groupby('sentiment')['engagement_rate'].mean().round(2)
```

# Task 5 — Statistical Analysis

Mean, Median, Mode

```python
# Mean
print(df[['likes','comments','shares','watch_time_sec','engagement_rate','follower_count']].mean().round(2))

# Median
print(df[['likes','comments','shares','watch_time_sec','engagement_rate','follower_count']].median().round(2))

# Mode
print(df[['likes','comments','shares','watch_time_sec','engagement_rate','follower_count']].mode().iloc[0].round(2))
```

Standard Deviation & Variance

```python
# Standard deviation
print(df[['likes','comments','shares','watch_time_sec','engagement_rate','follower_count']].std().round(2))

# Variance
print(df[['likes','comments','shares','watch_time_sec','engagement_rate','follower_count']].var().round(2))
```

Percentiles

```python
# 25th, 50th, 75th percentiles
print(df[['likes','comments','shares','watch_time_sec','engagement_rate','follower_count']].quantile([0.25,0.5,0.75]))
```

Skewness & Kurtosis

```python
# Skewness
print(df[['likes','comments','shares','watch_time_sec','engagement_rate','follower_count']].skew().round(2))

# Kurtosis
print(df[['likes','comments','shares','watch_time_sec','engagement_rate','follower_count']].kurt().round(2))
```
